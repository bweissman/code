```polyglot-notebook
IF EXISTS (SELECT * FROM sys.credentials  WHERE name = 'https://bwmanagedbackup.blob.core.windows.net/backup')
BEGIN
    DROP CREDENTIAL [https://bwmanagedbackup.blob.core.windows.net/backup]
END
```
